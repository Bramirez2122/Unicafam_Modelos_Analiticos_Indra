Consolidado Trimestral

```python
import pandas as pd

# Creamos un DataFrame de ejemplo 'consolidado' para que el código funcione
consolidado = pd.DataFrame({
    "sucursal": ["Norte", "Sur", "Norte", "Sur", "Norte", "Sur"],
    "mes": ["Ene", "Ene", "Mar", "Mar", "Ene", "Mar"],
    "ventas": [100, 150, 120, 180, 110, 190]
})

t = consolidado.pivot_table(index="sucursal", columns="mes",
                            values="ventas")
t["crecimiento_%"] = ((t["Mar"] - t["Ene"]) / t["Ene"] * 100).round(1)
t = t.sort_values("crecimiento_%", ascending=False)
print(t["crecimiento_%"])
print("Líder:", t.index[0])
```

SQL VS PANDAS

```python
import sqlite3
import pandas as pd

# Creamos una conexión SQLite en memoria para la demostración
con = sqlite3.connect(":memory:")

# Añadimos columnas de prueba a 'pedidos' para evitar el KeyError
pedidos["pago.medio"] = ["Tarjeta", "Efectivo", "Tarjeta"]
pedidos["pago.valor"] = [150.50, 89.90, 210.00]

# Tu código original
por_medio = pedidos.groupby("pago.medio")["pago.valor"].sum()
print("pandas:", por_medio.idxmax(), por_medio.max())
pedidos.rename(columns={"pago.medio": "medio",
                        "pago.valor": "valor"}).to_sql(
    "pagos", con, index=False, if_exists="replace")
q = """SELECT medio, SUM(valor) AS total FROM pagos
       GROUP BY medio ORDER BY total DESC LIMIT 1"""
print("SQL:", pd.read_sql(q, con).values.tolist())
```
